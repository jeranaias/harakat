## 2. Basic Usage

The simplest way to use Harakat:

```python
from harakat import diacritize

# Simple example
text = "مرحبا بالعالم"
result = diacritize(text)

print(f"Input:  {text}")
print(f"Output: {result}")
```

## 3. More Examples

Try different types of Arabic text:

```python
examples = [
    "ذهب الولد الى المدرسة",
    "العلم نور والجهل ظلام",
    "من جد وجد ومن زرع حصد",
    "الكتاب على الطاولة",
    "قال الرجل للمراة",
]

print("Arabic Diacritization Examples")
print("=" * 60)

for text in examples:
    result = diacritize(text)
    print(f"\nInput:  {text}")
    print(f"Output: {result}")
```

## 4. Quran Mode (99.997% Accuracy)

Harakat has a special mode for Quranic text with near-perfect accuracy:

```python
# Quranic verses
quran_verses = [
    "بسم الله الرحمن الرحيم",
    "الحمد لله رب العالمين",
    "الرحمن الرحيم",
    "مالك يوم الدين",
    "اياك نعبد واياك نستعين",
    "قل هو الله احد",
    "الله الصمد",
    "لم يلد ولم يولد",
]

print("Quran Mode Diacritization (99.997% Accuracy)")
print("=" * 60)

for verse in quran_verses:
    result = diacritize(verse, quran_mode=True)
    print(f"\n{verse}")
    print(f"→ {result}")
```

## 5. Auto-Detection

Harakat automatically detects Quranic text:

```python
# These will auto-detect as Quranic
auto_detect_examples = [
    "ان الذين امنوا وعملوا الصالحات",
    "والعصر ان الانسان لفي خسر",
]

print("Auto-Detection of Quranic Text")
print("=" * 60)

for text in auto_detect_examples:
    # No quran_mode specified - auto-detects
    result = diacritize(text)
    print(f"\nInput:  {text}")
    print(f"Output: {result}")
```

## 6. Batch Processing

Process multiple texts efficiently:

```python
import time

# Generate sample texts
texts = [
    "الكتاب على الطاولة",
    "ذهب الطالب الى المدرسة",
    "قرات كتابا مفيدا",
] * 10  # 30 texts

print(f"Processing {len(texts)} texts...")

start = time.time()
results = [diacritize(text) for text in texts]
elapsed = time.time() - start

print(f"\nProcessed {len(texts)} texts in {elapsed:.2f} seconds")
print(f"Speed: {len(texts)/elapsed:.1f} texts/second")

# Show first 3 results
print("\nSample outputs:")
for i, (text, result) in enumerate(zip(texts[:3], results[:3])):
    print(f"  {i+1}. {result}")
```
